# Exploratory Data Analysis (EDA)

```python
df = pd.read_csv("../data/cleaned_data.csv")
```

```python
import matplotlib.pyplot as plt
import plotly.express as px
```

```python
df.shape
```

```python
df.columns
```

```python
df.info()
```

Business KPI Dashboard

```python
print("Total Sales :", df["Sales"].sum())
print("Total Profit :", df["Gross Profit"].sum())
print("Total Orders :", df["Order ID"].nunique())
print("Total Products :", df["Product Name"].nunique())
print("Average Lead Time :", df["Lead Time"].mean())
```

Sales by Division

```python
division_sales = (
    df.groupby("Division")["Sales"]
      .sum()
      .sort_values(ascending=False)
)

division_sales
```

```python
fig = px.bar(
    division_sales,
    title="Sales by Division",
    labels={"value": "Sales", "Division": "Division"}
)

fig.show()
```

## Insight

- Chocolate division generates the highest sales.
- Sugar division contributes the least revenue.
- The company should focus on expanding chocolate products while improving the sales strategy for the sugar division.

Top 10 Products by Sales

```python
top_products = (
    df.groupby("Product Name")["Sales"]
      .sum()
      .sort_values(ascending=False)
      .head(10)
)

top_products
```

Top 10 Products on sale

```python
fig = px.bar(
    top_products,
    title="Top 10 Products by Sales"
)

fig.show()
```

## Insight

- Wonka Bar – Triple Dazzle Caramel is the highest-selling product.
- Most of the top-selling products belong to the Chocolate division.
- These products should be prioritized for production and inventory planning.

Sales by Region

```python
region_sales = (
    df.groupby("Region")["Sales"]
      .sum()
      .sort_values(ascending=False)
)

fig = px.pie(
    values=region_sales.values,
    names=region_sales.index,
    title="Sales Distribution by Region"
)

fig.show()
```

## Insight

- Pacific region contributes the highest sales.
- Gulf region contributes the lowest sales.
- The company should investigate demand in the Gulf region and improve its market presence.

Profit by Division

```python
profit = (
    df.groupby("Division")["Gross Profit"]
      .sum()
      .sort_values(ascending=False)
)

fig = px.bar(
    profit,
    title="Profit by Division"
)

fig.show()
```

## Insight

- Chocolate division contributes the highest profit.
- Sugar division generates minimal profit.
- Chocolate products should remain the company's primary focus for maximizing profitability.

Ship Mode Usage

```python
ship_mode = (
    df.groupby("Ship Mode")
      .size()
      .sort_values(ascending=False)
)

ship_mode
```

```python
fig = px.bar(
    ship_mode,
    title="Orders by Ship Mode"
)

fig.show()
```

## Insight

- Standard Class is the most frequently used shipping method.
- Same Day shipping is used the least.
- Most customers prefer economical shipping over faster delivery options.

Average Lead Time by Region

```python
lead_time = (
    df.groupby("Region")["Lead Time"]
      .mean()
      .sort_values(ascending=False)
)

lead_time
```

```python
fig = px.bar(
    lead_time,
    title="Average Lead Time by Region"
)

fig.show()
```

## Insight

- The average lead time is almost the same across all regions.
- No region experiences significantly higher delivery delays.
- This indicates a consistent shipping performance throughout the distribution network.

Top 10 Cities by Sales

```python
city_sales = (
    df.groupby("City")["Sales"]
      .sum()
      .sort_values(ascending=False)
      .head(10)
)

city_sales
```

```python
fig = px.bar(
    city_sales,
    title="Top 10 Cities by Sales"
)

fig.show()
```

## Insight

- New York City records the highest sales.
- Los Angeles is the second-highest contributing city.
- These cities should be prioritized for inventory allocation and marketing campaigns.

Monthly Sales Trend

```python
monthly_sales = (
    df.groupby("Order Month")["Sales"]
      .sum()
)

monthly_sales
```

```python
fig = px.line(
    monthly_sales,
    title="Monthly Sales Trend",
    markers=True
)

fig.show()
```

## Insight

- Sales gradually increase throughout the year.
- A significant increase is observed during the final quarter.
- Demand peaks in November and December, indicating seasonal sales growth.

Correlation Heatmap

```python
import plotly.express as px

corr = df[["Sales", "Gross Profit", "Cost", "Units", "Lead Time"]].corr()

fig = px.imshow(
    corr,
    text_auto=True,
    title="Correlation Heatmap"
)

fig.show()
```

## Insight

- Sales and Gross Profit have a very strong positive correlation.
- Sales and Cost are also highly correlated.
- Lead Time has very little correlation with sales and profit, suggesting delivery time does not significantly influence revenue.
